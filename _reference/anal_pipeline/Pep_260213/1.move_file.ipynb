```python
import os
import shutil
save_path = "./pep_data"
read_path = "artificial_peps"

file_type = ["IGH","IGL","IGK","TRA","TRB","TRG","TRD"]
for t in file_type:
    os.makedirs(os.path.join(save_path,t))

for dirname,dirs,filenames in os.walk(read_path):
    for filename in filenames:
        ftype = os.path.split(filename)[-1].split("__")[-1][:3]
        new_path = os.path.join(save_path,ftype)
        shutil.copy(os.path.join(dirname,filename),new_path)
```
