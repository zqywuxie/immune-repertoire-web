```python
import os
import pandas as pd
```

```python
file_paths = []
for dirname,dirs,filenames in os.walk("./1Vusage/"):
    for filename in filenames:
        file_paths.append(os.path.join(dirname,filename))
file_paths
```

```python
df_all = pd.DataFrame(columns=["sample","Category"])
for file_path in file_paths:
    df=  pd.read_csv(file_path)
    values = [sample_name.split("__")[0] for sample_name in df["sample"].tolist()]
    df = df.drop(columns=["sample"])
    df.insert(loc=0,column="sample",value=values)
    df_all = pd.merge(df_all,df,how="outer")
```

```python
df_all.to_csv("df_VJ_all.csv",index=False)
```
