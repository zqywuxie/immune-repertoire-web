```python
import sklearn
import os
import pandas as pd
```

```python
df = pd.read_csv("Profile_All.csv")
df = df[~df["therapy"].isna()]
df.to_csv("Profile_All_procced.csv", index=False)   
```

```python
df_dp = pd.read_csv("./data/Profile_1.csv")
df_dp = df_dp[["sample","Group1"]+df_dp.columns[df_dp.columns.to_list().index("TRA_percent_reads_all"):].tolist()]
```

```python
for dirname,dirs,filenames in os.walk("./1VJusage"):
    for filename in filenames:
        df = pd.read_csv(os.path.join(dirname,filename))
        df = df.drop(columns="Category")
        df["sample"] = df["sample"].str.split("__").str[0]
        df_dp = pd.merge(df_dp,df,on="sample")
```

```python
df_dp.to_csv("merged.csv",index=False)
```
